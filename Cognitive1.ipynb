Assignment 5.3: WAP to add all prime numbers from 1 to n, n is given by the user.

```python
import math
n = int(input('enter the no. : '))
s = 0
for i in range(1,n+1):
  g = 1
  for j in range(2,int(math.sqrt(i))+1):
    if i % j == 0:
      g = 0
      break
  if g == 1 and i > 1:
    s = s + i

print('sum of all prime numbers from 1 to',n,'is:',s)
```

Assignment 6.1: WAP using a function that adds all odd numbers from 1 to n, n is given by the user.

```python
def sum_of_odds(n):
  s = 0
  for i in range(1, n+1):
    if i % 2 != 0:
      s = s + i
  return s

n = int(input('enter the number: '))
print('sum of all odd number up to', n, 'is:', sum_of_odds(n))
```

Assignment 6.2: WAP using a function that adds all prime numbers from 1 to n, n given by the user.


```python
import math

def is_prime(num):
    if num < 2:
        return False

    for j in range(2, int(math.sqrt(num)) + 1):
        if num % j == 0:
            return False

    return True
def sum_of_primes(n):
    s = 0

    for i in range(1, n + 1):
        if is_prime(i):
            s = s + i

    return s
n = int(input("Enter the number: "))
print("Sum of all prime numbers from 1 to", n, "is:", sum_of_primes(n))
```
